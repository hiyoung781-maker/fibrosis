```python
import glob
import pandas as pd

# 전체 파티션 로드 (id, approvedSymbol, tractability만 필요)
target_dir = '/home/seyoung/academic_symposium/fibrosis/opentarget/target'
parquet_files = sorted(glob.glob(f'{target_dir}/part-*.snappy.parquet'))
df_tract = pd.concat(
    [pd.read_parquet(p, columns=['id', 'approvedSymbol', 'tractability']) for p in parquet_files],
    ignore_index=True,
)

# Small Molecule(SM) tractability 카테고리 -> A/B/C 등급
# 우선순위: A(임상 선례) > B(우리 사냥터) > C(약한 근거)
GRADE_MAP = {
    'Approved Drug':          'A',
    'Advanced Clinical':      'A',
    'Phase 1 Clinical':       'A',
    'Structure with Ligand':  'B',
    'High-Quality Ligand':    'B',
    'High-Quality Pocket':    'B',
    'Med-Quality Pocket':     'C',
    'Druggable Family':       'C',
}
GRADE_ORDER = {'A': 0, 'B': 1, 'C': 2}


def sm_tractability_grade(tractability):
    """SM modality 항목 중 value=True인 카테고리들에서 가장 높은 등급(A>B>C)을 반환. 해당 없으면 None."""
    if tractability is None:
        return None
    grades = [
        GRADE_MAP[row['id']]
        for row in tractability
        if row['modality'] == 'SM' and row['value'] and row['id'] in GRADE_MAP
    ]
    return min(grades, key=lambda g: GRADE_ORDER[g]) if grades else None


df_tract['sm_tractability_grade'] = df_tract['tractability'].apply(sm_tractability_grade)

print("등급별 유전자 수:")
print(df_tract['sm_tractability_grade'].value_counts(dropna=False))

print("\n등급 A(임상 선례) 예시:")
df_tract[df_tract['sm_tractability_grade'] == 'A'][['id', 'approvedSymbol', 'sm_tractability_grade']].head(10)
```
